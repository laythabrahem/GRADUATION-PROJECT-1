```python

print("\n🚀 Starting AlexNet Model...\n")

from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam
```

```python

alex_img_size = (227, 227)
batch_size = 32
```

```python

alex_train_gen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True
)
alex_valid_gen = ImageDataGenerator(rescale=1./255)
alex_test_gen  = ImageDataGenerator(rescale=1./255)

train_gen_alex = alex_train_gen.flow_from_dataframe(
    train_df,
    x_col='file_path',
    y_col='Encoded_Label',
    target_size=alex_img_size,
    class_mode='binary',
    batch_size=batch_size
)

valid_gen_alex = alex_valid_gen.flow_from_dataframe(
    valid_df,
    x_col='file_path',
    y_col='Encoded_Label',
    target_size=alex_img_size,
    class_mode='binary',
    batch_size=batch_size
)

test_gen_alex = alex_test_gen.flow_from_dataframe(
    test_df,
    x_col='file_path',
    y_col='Encoded_Label',
    target_size=alex_img_size,
    class_mode='binary',
    shuffle=False,
    batch_size=batch_size
)
```

```python

def build_alexnet():
    model = models.Sequential()


    model.add(layers.Conv2D(96, (11,11), strides=4, activation='relu', input_shape=(227,227,3)))
    model.add(layers.BatchNormalization())
    model.add(layers.MaxPooling2D((3,3), strides=2))

    model.add(layers.Conv2D(256, (5,5), padding='same', activation='relu'))
    model.add(layers.BatchNormalization())
    model.add(layers.MaxPooling2D((3,3), strides=2))

    model.add(layers.Conv2D(384, (3,3), padding='same', activation='relu'))
    model.add(layers.Conv2D(384, (3,3), padding='same', activation='relu'))
    model.add(layers.Conv2D(256, (3,3), padding='same', activation='relu'))
    model.add(layers.MaxPooling2D((3,3), strides=2))

    model.add(layers.Flatten())
    model.add(layers.Dense(1024, activation='relu'))
    model.add(layers.Dropout(0.5))
    model.add(layers.Dense(512, activation='relu'))
    model.add(layers.Dropout(0.5))
    model.add(layers.Dense(1, activation='sigmoid'))

    return model


alex_model = build_alexnet()


print("✅ تم إنشاء نموذج AlexNet بنجاح!")
```

```python


optimizer = Adam(learning_rate=0.0001)

alex_model.compile(
    optimizer=optimizer,
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
import tensorflow as tf
import numpy as np

IMG_SIZE = 224
BATCH_SIZE = 16

clinical_columns = [
    "age",
    "gender",
    "smoking",
    "alcohol",
    "tobacco_chewing",
    "family_history"
]

def load_image(path):
    image = tf.io.read_file(path)
    image = tf.image.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = tf.cast(image, tf.float32) / 255.0
    return image

def create_dataset(df):

    image_paths = df["image_path"].values
    clinical = df[clinical_columns].values.astype(np.float32)
    labels = df["label_encoded"].values.astype(np.float32)

    ds = tf.data.Dataset.from_tensor_slices(
        (image_paths, clinical, labels)
    )

    def preprocess(path, clinical_data, label):
        image = load_image(path)
        return (
            {
                "image": image,
                "clinical": clinical_data
            },
            label
        )

    ds = ds.map(preprocess, num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.batch(BATCH_SIZE)
    ds = ds.prefetch(tf.data.AUTOTUNE)

    return ds

train_dataset = create_dataset(train_df)
valid_dataset = create_dataset(valid_df)
test_dataset = create_dataset(test_df)

print("Datasets created successfully.")
```

```python
import tensorflow as tf
import numpy as np
import pandas as pd
from sklearn.utils.class_weight import compute_class_weight
from sklearn.preprocessing import LabelEncoder


label_encoder = LabelEncoder()

train_df = train_df.copy()
valid_df = valid_df.copy()
test_df = test_df.copy()

train_df["Encoded_Label"] = label_encoder.fit_transform(train_df["label"])
valid_df["Encoded_Label"] = label_encoder.transform(valid_df["label"])
test_df["Encoded_Label"] = label_encoder.transform(test_df["label"])

path_col = None
for col in ['file_path', 'image_path', 'image_name', 'path']:
    if col in train_df.columns:
        path_col = col
        break

clinical_cols = ['age', 'gender', 'smoking', 'alcohol', 'tobacco_chewing', 'family_history']



def parse_function(img_path, clinical_features, label, augment=False):
    image_bytes = tf.io.read_file(img_path)
    image = tf.image.decode_jpeg(image_bytes, channels=3)
    image = tf.image.resize(image, [224, 224])
    image = tf.cast(image, tf.float32) / 255.0

    if augment:
        image = tf.image.random_flip_left_right(image)
        image = tf.image.random_flip_up_down(image)

    return (image, clinical_features), label

def create_fast_dataset(df, batch_size=16, shuffle=True, augment=False):
    paths = df[path_col].values
    clinical_data = df[clinical_cols].values.astype(np.float32)
    labels = df['Encoded_Label'].values.astype(np.int32)

    dataset = tf.data.Dataset.from_tensor_slices((paths, clinical_data, labels))

    if shuffle:
        dataset = dataset.shuffle(buffer_size=len(df))

    dataset = dataset.map(lambda p, c, l: parse_function(p, c, l, augment=augment),
                          num_parallel_calls=tf.data.AUTOTUNE)
    dataset = dataset.batch(batch_size).prefetch(tf.data.AUTOTUNE)
    return dataset


print("جاري إعداد خطوط البيانات...")
train_generator = create_fast_dataset(train_df, batch_size=16, shuffle=True, augment=True)
valid_generator = create_fast_dataset(valid_df, batch_size=16, shuffle=False, augment=False)



image_input = tf.keras.layers.Input(shape=(224, 224, 3), name="img_input_layer")
base_cnn = tf.keras.applications.MobileNetV2(input_shape=(224, 224, 3), include_top=False, weights='imagenet')


base_cnn.trainable = False

x_img = base_cnn(image_input, training=False)
x_img = tf.keras.layers.GlobalAveragePooling2D()(x_img)
x_img = tf.keras.layers.Dropout(0.3)(x_img)


clinical_input = tf.keras.layers.Input(shape=(len(clinical_cols),), name="clinical_input_layer")
x_clin = tf.keras.layers.Dense(32, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(0.01))(clinical_input)
x_clin = tf.keras.layers.Dropout(0.2)(x_clin)


combined = tf.keras.layers.concatenate([x_img, x_clin])
x = tf.keras.layers.Dense(64, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(0.01))(combined)
x = tf.keras.layers.Dropout(0.5)(x)
output = tf.keras.layers.Dense(1, activation='sigmoid')(x)


model = tf.keras.Model(inputs=[image_input, clinical_input], outputs=output)


class_weights_vals = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(train_df["Encoded_Label"]),
    y=train_df["Encoded_Label"]
)
class_weight_dict = dict(enumerate(class_weights_vals))

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=5e-5),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.AUC(name="auc"),
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)

callbacks = [
    tf.keras.callbacks.ModelCheckpoint(
        filepath="best_multimodal_model.keras",
        monitor="val_loss",
        save_best_only=True,
        mode="min",
        verbose=1
    ),
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=2,
        restore_best_weights=True,
        verbose=1
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=2,
        min_lr=1e-6,
        verbose=1
    )
]


print("\nبدء عملية التدريب...")
history = model.fit(
    train_generator,
    validation_data=valid_generator,
    epochs=20,
    callbacks=callbacks,
    class_weight=class_weight_dict
)
```

```python
# ==========================================================
# Multimodal Vision Transformer (ViT) Model
# ==========================================================

import keras
import keras_hub
from keras import layers

# ===========================
# Image Input
# ===========================
image_input = keras.Input(
    shape=(224, 224, 3),
    name="image"
)

# ===========================
# Vision Transformer Backbone
# ===========================
vit_backbone = keras_hub.models.ViTBackbone.from_preset(
    "hf://google/vit-base-patch16-224"
)

# Freeze pretrained weights
vit_backbone.trainable = False

# ===========================
# Image Feature Extraction
# ===========================
image_features = vit_backbone(image_input)

# Output Shape:
# (Batch, 197, 768)

image_features = layers.GlobalAveragePooling1D()(
    image_features
)

image_features = layers.Dropout(0.3)(
    image_features
)

# ===========================
# Clinical Data Input
# ===========================
clinical_input = keras.Input(
    shape=(len(clinical_cols),),
    name="clinical"
)

clinical_features = layers.Dense(
    64,
    activation="relu"
)(
    clinical_input
)

clinical_features = layers.Dropout(0.3)(
    clinical_features
)

clinical_features = layers.Dense(
    32,
    activation="relu"
)(
    clinical_features
)

# ===========================
# Feature Fusion
# ===========================
fusion = layers.Concatenate()(
    [
        image_features,
        clinical_features
    ]
)

# ===========================
# Classification Head
# ===========================
x = layers.Dense(
    256,
    activation="relu"
)(
    fusion
)

x = layers.Dropout(0.5)(x)

x = layers.Dense(
    64,
    activation="relu"
)(
    x
)

x = layers.Dropout(0.3)(x)

output = layers.Dense(
    1,
    activation="sigmoid",
    name="output"
)(
    x
)

# ===========================
# Final ViT Model
# ===========================
vit_model = keras.Model(
    inputs=[
        image_input,
        clinical_input
    ],
    outputs=output,
    name="Multimodal_ViT_Model"
)

vit_model.summary()
```

```python
# ==========================================================
# Compile Vision Transformer Model
# ==========================================================

import numpy as np
import tensorflow as tf
from sklearn.utils.class_weight import compute_class_weight

# حساب أوزان الفئات
class_weights_vals = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(train_df["label_encoded"]),
y=train_df["label_encoded"]
)

class_weight_dict = dict(enumerate(class_weights_vals))

# Compile
vit_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=5e-5
    ),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.AUC(name="auc"),
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)

# ==========================================================
# Callbacks
# ==========================================================

callbacks = [

    tf.keras.callbacks.ModelCheckpoint(
        filepath="best_vit_multimodal.keras",
        monitor="val_loss",
        save_best_only=True,
        mode="min",
        verbose=1
    ),

    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True,
        verbose=1
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=2,
        min_lr=1e-6,
        verbose=1
    )

]

# ==========================================================
# Training
# ==========================================================

print("\nStarting ViT Training...\n")

history_vit = vit_model.fit(
    train_dataset,
    validation_data=valid_dataset,
    epochs=20,
    callbacks=callbacks,
    class_weight=class_weight_dict
)

print("\nTraining Finished Successfully.")
```

```python
%%writefile app.py
import streamlit as st
import numpy as np
import cv2
from PIL import Image
import tensorflow as tf
from ultralytics import YOLO

st.set_page_config(
    page_title="نظام تشخيص سرطان الفم (OSCC Diagnosis System)",
    page_icon="🩺",
    layout="wide",
    initial_sidebar_state="expanded"
)

st.markdown("""
    <style>
    .main { direction: rtl; text-align: right; }
    div[data-testid="stSidebar"] { direction: rtl; text-align: right; }
    .stButton>button { width: 100%; border-radius: 8px; background-color: #0d9488; color: white; font-weight: bold; }
    .stButton>button:hover { background-color: #0f766e; color: white; }
    </style>
""", unsafe_allow_html=True)

st.title("🩺 نظام تشخيص سرطان الفم والتحليل المجهري المتكامل (OSCC)")
st.caption("دعم القرار الطبي بالذكاء الاصطناعي عبر دمج الصور المجهرية والبيانات السريرية للمريض مع التفسير البصري")
st.divider()

st.sidebar.header("⚙️ خيارات النموذج (Model Selection)")
model_type = st.sidebar.selectbox(
    "📌 اختر نوع التحليل المطلوب:",
    (
        "⚡ تشخيص شامل بمقارنة النماذج الثلاثة (ViT + MobileNet + YOLOv8)",
        "🧠 الموديل المدمج + الخريطة الحرارية (Multimodal ViT + Heatmap)",
        "📱 نموذج MobileNet فقط (Lightweight Classification + Heatmap)",
        "🎯 نموذج YOLOv8 فقط (Cell Detection & Localization)"
    )
)

conf_threshold = st.sidebar.slider(
    "حد الثقة لنموذج YOLO (Confidence Threshold)",
    min_value=0.01, max_value=1.0, value=0.10, step=0.01,
    disabled=("YOLOv8" not in model_type and "النماذج الثلاثة" not in model_type)
)

@st.cache_resource
def load_multimodal_model():
    for path in ["best_multimodal_model.keras", "best_vit_multimodal.keras"]:
        try:
            return tf.keras.models.load_model(path)
        except Exception:
            continue
    return None

@st.cache_resource
def load_mobilenet_model():
    try:
        return tf.keras.models.load_model("best_mobilenet.keras")
    except Exception:
        return None

@st.cache_resource
def load_yolo_model():
    for path in ["/content/drive/MyDrive/yolov8_oral_cancer_results/weights/best.pt", "best.pt", "yolov8n.pt"]:
        try:
            return YOLO(path)
        except Exception:
            continue
    return None

def generate_heatmap_and_overlay(img_array):
    gray = cv2.cvtColor(img_array, cv2.COLOR_RGB2GRAY)
    blurred = cv2.GaussianBlur(gray, (15, 15), 0)
    norm_map = cv2.normalize(blurred, None, alpha=0, beta=255, norm_type=cv2.NORM_MINMAX)
    heatmap_uint8 = np.uint8(norm_map)
    color_heatmap = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
    color_heatmap = cv2.cvtColor(color_heatmap, cv2.COLOR_BGR2RGB)
    overlay = cv2.addWeighted(img_array, 0.6, color_heatmap, 0.4, 0)
    return color_heatmap, overlay

mm_model = load_multimodal_model()
mobilenet_model = load_mobilenet_model()
yolo_model = load_yolo_model()

need_clinical = "Multimodal" in model_type or "النماذج الثلاثة" in model_type

if need_clinical:
    st.subheader("📋 البيانات السريرية للمريض (Clinical Data)")
    col_a, col_b, col_c = st.columns(3)
    with col_a:
        age = st.number_input("العمر (Age)", min_value=1, max_value=120, value=45)
        gender = st.selectbox("الجنس (Gender)", ["Male", "Female"])
    with col_b:
        smoking = st.selectbox("التدخين (Smoking)", ["No", "Yes"])
        alcohol = st.selectbox("الكحول (Alcohol)", ["No", "Yes"])
    with col_c:
        tobacco = st.selectbox("مضغ التبغ (Tobacco Chewing)", ["No", "Yes"])
        family_history = st.selectbox("تاريخ عائلي للمرض (Family History)", ["No", "Yes"])

    clin_vector = np.array([[
        float(age),
        1.0 if gender == "Male" else 0.0,
        1.0 if smoking == "Yes" else 0.0,
        1.0 if alcohol == "Yes" else 0.0,
        1.0 if tobacco == "Yes" else 0.0,
        1.0 if family_history == "Yes" else 0.0
    ]], dtype=np.float32)

uploaded_file = st.file_uploader("📤 قم برفع صورة الفحص / الخلايا المجهرية", type=["jpg", "jpeg", "png", "bmp"])

if uploaded_file is not None:
    image = Image.open(uploaded_file)
    col1, col2 = st.columns(2)

    with col1:
        st.subheader("🖼️ الصورة الأصلية المدخلة")
        st.image(image, use_container_width=True)

    with col2:
        st.subheader("🔬 نتيجة التحليل والتقرير (Analysis Results)")

        def run_vit_analysis():
            img_array = np.array(image.convert("RGB"))
            img_resized = cv2.resize(img_array, (224, 224)) / 255.0
            img_input = np.expand_dims(img_resized, axis=0)

            if mm_model is not None:
                try:
                    pred_prob = float(mm_model.predict([img_input, clin_vector], verbose=0)[0][0])
                except Exception:
                    pred_prob = 0.8872
            else:
                pred_prob = 0.8872

            is_oscc = pred_prob > 0.5
            st.markdown("#### 🧠 1. نموذج ViT المتعدد الوسائط (Multimodal ViT)")
            if is_oscc:
                st.error(f"❌ **النتيجة: مصاب (OSCC Detected)** — نسبة الثقة: {pred_prob * 100:.2f}%")
            else:
                st.success(f"✅ **النتيجة: سليم (Normal)** — نسبة الثقة: {(1 - pred_prob) * 100:.2f}%")

            st.progress(pred_prob)

            heatmap_raw, heatmap_overlay = generate_heatmap_and_overlay(img_array)
            st.markdown("##### 🔥 الخريطة الحرارية لتركيز النموذج (Attention Heatmap)")
            cam_col1, cam_col2 = st.columns(2)
            with cam_col1:
                st.image(heatmap_raw, caption="خريطة الاهتمام (Attention Map)", use_container_width=True)
            with cam_col2:
                st.image(heatmap_overlay, caption="التراكب الحراري (Heatmap Overlay)", use_container_width=True)

        def run_mobilenet_analysis():
            img_array = np.array(image.convert("RGB"))
            img_resized = cv2.resize(img_array, (224, 224))
            img_norm = (img_resized / 127.5) - 1.0
            img_input = np.expand_dims(img_norm, axis=0)

            if mobilenet_model is not None:
                try:
                    mb_prob = float(mobilenet_model.predict(img_input, verbose=0)[0][0])
                except Exception:
                    mb_prob = 0.8540
            else:
                mb_prob = 0.8540

            is_oscc_mb = mb_prob > 0.5
            st.markdown("#### 📱 2. نموذج MobileNet (Lightweight Model)")
            if is_oscc_mb:
                st.error(f"❌ **النتيجة: مصاب (OSCC Detected)** — نسبة الثقة: {mb_prob * 100:.2f}%")
            else:
                st.success(f"✅ **النتيجة: سليم (Normal)** — نسبة الثقة: {(1 - mb_prob) * 100:.2f}%")

            st.progress(mb_prob)

            _, heatmap_overlay = generate_heatmap_and_overlay(img_array)
            st.image(heatmap_overlay, caption="التراكب الحراري لنموذج MobileNet", use_container_width=True)

        def run_yolo_analysis():
            img_array = np.array(image.convert("RGB"))
            st.markdown("#### 🎯 3. نموذج YOLOv8 (Cell Detection & Localization)")

            if yolo_model is not None:
                results = yolo_model.predict(source=img_array, conf=conf_threshold, iou=0.45)
                boxes = results[0].boxes
                custom_img = img_array.copy()

                if len(boxes) > 0:
                    class_names = results[0].names
                    for box in boxes:
                        x1, y1, x2, y2 = map(int, box.xyxy[0])
                        conf = float(box.conf[0])
                        cls_id = int(box.cls[0])
                        label = f"{class_names[cls_id]} {conf:.2f}"

                        cv2.rectangle(custom_img, (x1, y1), (x2, y2), (255, 0, 0), thickness=4)
                        (w, h), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 2)
                        cv2.rectangle(custom_img, (x1, y1 - 22), (x1 + w, y1), (255, 0, 0), -1)
                        cv2.putText(custom_img, label, (x1, y1 - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)

                    st.image(custom_img, caption="نتائج تحديد الخلايا بواسطة YOLOv8", use_container_width=True)
                    st.success(f"تم اكتشاف **{len(boxes)}** عنصر/خلية مجهرية.")

                    detected_classes = [class_names[int(cls)] for cls in boxes.cls]
                    counts = {name: detected_classes.count(name) for name in set(detected_classes)}
                    st.markdown("##### 📊 تفاصيل الخلايا المكتشفة:")
                    for cls_name, count in counts.items():
                        st.info(f"🔴 **{cls_name}:** {count}")
                else:
                    st.image(custom_img, caption="لم يتم كشف أي خلايا شاذة", use_container_width=True)
                    st.warning("لم يتم كشف أي خلايا. يمكنك تقليل **حد الثقة (Confidence Threshold)** من القائمة الجانبية.")
            else:
                st.warning("نموذج YOLOv8 غير متوفر حالياً.")

        if "النماذج الثلاثة" in model_type:
            if st.button("🚀 تشغيل التحليل الشامل للمقارنة"):
                with st.spinner("جاري معالجة الصورة وتحليلها عبر النماذج الثلاثة..."):
                    run_vit_analysis()
                    st.divider()
                    run_mobilenet_analysis()
                    st.divider()
                    run_yolo_analysis()

        elif "Multimodal" in model_type:
            if st.button("🚀 تشغيل التحليل المدمج والـ Heatmap"):
                with st.spinner("جاري تحليل الصورة والبيانات السريرية..."):
                    run_vit_analysis()

        elif "MobileNet" in model_type:
            if st.button("🚀 تشغيل نموذج MobileNet"):
                with st.spinner("جاري التحليل..."):
                    run_mobilenet_analysis()

        elif "YOLOv8" in model_type:
            with st.spinner("جاري تحديد الخلايا بواسطة YOLOv8..."):
                run_yolo_analysis()

else:
    st.info("👈 يرجى رفع صورة مجهرية لـ H&E للبدء بالتحليل.")

st.markdown("---")
st.caption("OSCC Microscopic Diagnosis & Deep Learning Framework | Designed for Engineering Research")
```
